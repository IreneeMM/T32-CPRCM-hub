# Check whether variables exist in the folder structure

The library Path is imported to work with directories

```python
#imports:
from pathlib import Path
```

The next cell search in a given directory (base_dir) with a given  internal structure (here "*/*/*/*/*/*/*/1hr") where there are a given folder, in the case that has the variable name, if it exist.

The following function searches for a given variable within the folder structure of the data directory specified in base_dir.

Starting from base_dir, it finds all 1hr folders that follow the expected internal structure: '*/*/*/*/*/*/*/1hr'

For each 1hr folder, the function checks whether it contains a subfolder with the requested variable name. It then reports how many folders contain that variable and how many are missing it.

```python
# Check whether a variable exists in the 1hr folders

def find_variable(variable):
    base_dir = Path("../I4C_CPRCM_DATA")
    missing_variable = []
    found_variable = []

    for hourly_dir in base_dir.glob("*/*/*/*/*/*/*/1hr"):
        variable_path = hourly_dir / variable

        if variable_path.is_dir():
            found_variable.append(hourly_dir)
        else:
            missing_variable.append(hourly_dir)

    print(f"Number of 1hr folders checked: {len(found_variable) + len(missing_variable)}")
    print(f"Number of 1hr folders with {variable} : {len(found_variable)}")
    print(f"Number of 1hr folders missing {variable} : {len(missing_variable)}")

    print("\n1hr folders where ", variable,  " is missing:")
    for folder in missing_variable:
        print(folder)
```

Below are usage examples for three different variable names: `tas`, `huss`, and `hurs`.


```python
# Check that tas exist in 1hr folders:
find_variable('tas')
```

```python
# Check that huss exist in 1hr folders:
find_variable('huss')
```

```python
# Check that huss exist in 1hr folders:
find_variable('hurs')
```

No directory is printed because the variable was found in all cases.
