# NetCDF Data Catalogue and Available Options
This notebook scans the NetCDF data directory, extracts the information contained in each directory level, creates a catalogue of the available files, and displays the unique options available for each parameter.

### Libraries required to run the script:

```python
from pathlib import Path
import pandas as pd
import os
import openpyxl
```

### Variables:

```python
# Substitute the string with your data directory
data_directory = Path("../I4C_CPRCM_DATA")

# Define one column for each directory level
columns = ["domain", "institution", "driving_model", "scenario", "member",
           "model", "version", "time_period", "variable", "data_version"]

# Store the information extracted from each file
records = []
```

### NetCDF File Catalogue Creation

This section recursively searches for NetCDF files, extracts the information from each directory level, and stores it together with the file name. Only files matching the expected directory structure are included in the catalogue, which is then converted into a pandas DataFrame.

```python
# Search for all NetCDF files and extract the directory information
for file in data_directory.rglob("*.nc"):#
    relative_parts = file.relative_to(data_directory).parts

    # There must be one folder per column plus the file name
    if len(relative_parts) == len(columns) + 1:
        record = dict(zip(columns, relative_parts))# Pair each column with its corresponding value in the path
        records.append(record)#Store the information for each file as an item in a list

# Create the catalogue
catalogue = pd.DataFrame(records)#Convert the records list of dictionaries into a pandas DataFrame
catalogue
```

### Available Options Extraction

This section extracts the unique values available for each catalogue parameter, removes missing values, and stores the results as alphabetically sorted lists. The available options are then displayed as comma-separated strings for easier consultation.

```python
# Store the unique available options as sorted lists
available_options = {column: sorted(catalogue[column].dropna().unique().tolist()) for column in columns}

# Display the 
for column, options in available_options.items():
    print(f"{column}: {', '.join(options)}")
```
